```python
import pandas as pd
from retailmind.data import enrich
from retailmind.forecast import backtest, daily_series
```

```python
data = enrich(pd.read_csv('../data/demo_sales.csv'))
results = backtest(daily_series(data, 'P001'), horizon=14)
results.groupby('model')[['WAPE', 'MAE', 'Bias', 'Underforecast', 'Overforecast']].mean().sort_values('WAPE').round(2)
```
