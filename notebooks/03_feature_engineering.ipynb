# Leakage-safe features

Run from the `notebooks/` directory after installing `requirements.txt`. Outputs are calculated from the bundled observed M5 subset.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
```

```python
import pandas as pd
from retailmind.data import enrich
from retailmind.forecast import daily_series, make_features
```

```python
data = enrich(pd.read_csv('../data/m5_observed.csv.gz', low_memory=False))
series = daily_series(data, data.product_id.iloc[0])
features = make_features(series)
features.tail()
```

```python
# The final row's demand-derived values depend on earlier rows only.
assert features.iloc[-1].lag_1 == series.iloc[-2]
```
