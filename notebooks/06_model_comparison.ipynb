# Model comparison and business consequences

Run from the `notebooks/` directory after installing `requirements.txt`. Outputs are calculated from the bundled observed M5 subset.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
```

```python
import pandas as pd
from retailmind.data import enrich
from retailmind.forecast import backtest, daily_series
```

```python
data = enrich(pd.read_csv('../data/m5_observed.csv.gz', low_memory=False))
results = backtest(daily_series(data, data.product_id.iloc[0]), horizon=14)
results.groupby('model')[['WAPE', 'MAE', 'Bias', 'Underforecast', 'Overforecast']].mean().sort_values('WAPE').round(2)
```
