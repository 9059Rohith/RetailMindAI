# Rolling-origin forecasting

Run from the `notebooks/` directory after installing `requirements.txt`. Outputs are calculated from the bundled observed M5 subset.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
```

```python
import pandas as pd
import plotly.express as px
from retailmind.data import enrich
from retailmind.forecast import daily_series, run_forecast
```

```python
data = enrich(pd.read_csv('../data/m5_observed.csv.gz', low_memory=False))
result = run_forecast(daily_series(data, data.product_id.iloc[0]), horizon=14)
result.comparison.round(2)
```

```python
px.line(result.future, x='date', y=['forecast', 'lower', 'upper'], title='Forecast and heuristic range')
```
